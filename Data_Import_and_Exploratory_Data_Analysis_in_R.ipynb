Calculate median, Q1 and Q3 and present minimum and maximum MonthlyIncome for each JobLevel

```r
IBM_DATA %>%
  group_by(JobLevel) %>%
  summarise(
    Minimo = min(MonthlyIncome),
    Q1 = quantile(MonthlyIncome, 0.25),
    Median = quantile(MonthlyIncome, 0.5),
    Q3 = quantile(MonthlyIncome, 0.75),
    Maximo = max(MonthlyIncome)
  )
```

Global quartiles of MonthlyIncome

```r
Q1 <- quantile(IBM_DATA$MonthlyIncome, 0.25)
Q3 <- quantile(IBM_DATA$MonthlyIncome, 0.75)
IQR_val <- IQR(IBM_DATA$MonthlyIncome)
```

```r
Q1
```

```r
Q3
```

MonthlyIncome Outliers: which ones, how many, and what percentage do they represent?

```r
outliersMonthlyIncome <- IBM_DATA$MonthlyIncome[
  IBM_DATA$MonthlyIncome < (Q1 - 1.5*IQR_val) |
  IBM_DATA$MonthlyIncome > (Q3 + 1.5*IQR_val)
]
```

```r
outliersMonthlyIncome
```

```r
length(outliersMonthlyIncome)
```

```r
round((length(outliersMonthlyIncome) / nrow(IBM_DATA)) * 100, 2)
```
